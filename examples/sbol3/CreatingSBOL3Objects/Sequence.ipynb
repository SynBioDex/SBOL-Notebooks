# Creating a Sequence

In the following tutorial, we will be creating a `Sequence` object.

`Sequence` objects have the following properties:
- `elements` specifies the make-up of the sequence (in this case, it is a string of nucleotide bases)
- `encoding` specifies how the sequence should be interpreted

For more information on the `Sequence` class and its properties, check out page 16 of the SBOL 3.1 specifications which can be found at the following [link](https://sbolstandard.org/docs/SBOL3.1.0.pdf).

Import the module

```python
import sbol3
```

Create the document and set the namespace

```python
doc = sbol3.Document()
sbol3.set_namespace('https://github.com/SynBioDex/SBOL-Notebooks')
```

Create a `Sequence` object and add it to the document

```python
seq = sbol3.Sequence(identity='example_sequence', 
                    elements='ATGCGTACGTAGCTAGCTAGC', 
                    encoding=sbol3.IUPAC_DNA_ENCODING)
doc.add(seq)
```

Validate the document

```python
for obj in doc.objects:
    print(obj.identity)
report_sbol3 = doc.validate()
print(len(report_sbol3))
```

Save the document to an SBOL file

```python
doc.write('../CreatedSBOL3Objects/sequence_example.xml')
```

```python
# Write document as sorted N-Triples (good for version control)
doc.write('../CreatedSBOL3Objects/sequence_example.nt', sbol3.SORTED_NTRIPLES)
```
