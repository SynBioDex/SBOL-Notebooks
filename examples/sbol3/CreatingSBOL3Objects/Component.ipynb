# Creating a Component

`Component` objects have the following properties:
- `identity`
- `types`
- `roles`
- `sequences`
- `features`
- `constraints`
- `interactions`
- `interface`
- `models`
- `attachments`

In this tutorial, we will be representing a ribosome binding site by creating a `Component` with a `Sequence`. To do this, we will only need to set the `identity`, `types`, `roles`, and `sequences` properties. For a guide on setting the other properties, check out the cooresponding notebooks.

For more information on the `Component` class and its properties, check out page 18 of the SBOL 3.1 specifications which can be found at the following [link](https://sbolstandard.org/docs/SBOL3.1.0.pdf).


Import the module

```python
import sbol3
```

Create the document and set the namespace

```python
doc = sbol3.Document()
sbol3.set_namespace('https://github.com/SynBioDex/SBOL-Notebooks')
```

Create a `Sequence` object and add it to the document

```python
seq = sbol3.Sequence(identity='example_sequence', 
                    elements='ATGCGTACGTAGCTAGCTAGC', 
                    encoding=sbol3.IUPAC_DNA_ENCODING
                    )
doc.add(seq)
```

Ceate a `Component` object for DNA encoding a ribosome binding site and add it to the document.

```python
component = sbol3.Component(identity='example_component', 
                            types=[sbol3.SBO_DNA],
                            roles=[sbol3.SO_RBS],
                            sequences=seq
                           )
doc.add(component)
```

Validate the document

```python
for obj in doc.objects:
    print(obj.identity)
report_sbol3 = doc.validate()
print(len(report_sbol3))
```

Save the document to an SBOL file

```python
doc.write('../CreatedSBOL3Objects/component_example.xml')
```

```python
# Write document as sorted N-Triples (good for version control)
doc.write('../CreatedSBOL3Objects/component_example.nt', sbol3.SORTED_NTRIPLES)
```
