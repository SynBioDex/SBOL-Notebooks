# Creating a Collection

In the following tutorial, we will be creating a `Collection` object.

`Collection` objects have the following properties:
- `members` Zero or more member properties, each of type IRI specifying a TopLevel object.


The Collection class is a class that groups together a set of TopLevel objects that have something in common.

For more information on the `Collection` class and its properties, check out page 38 of the SBOL 3.1 specifications which can be found at the following [link](https://sbolstandard.org/docs/SBOL3.1.0.pdf).

Import the module

```python
import sbol3
```

Create the document and set the namespace

```python
doc = sbol3.Document()
sbol3.set_namespace('https://github.com/SynBioDex/SBOL-Notebooks')
```

Create some `TopLevel` objects

```python
pro1 = sbol3.Component(identity='promoter_1', types=[sbol3.SBO_DNA], roles=[sbol3.SO_PROMOTER])
pro2 = sbol3.Component(identity='promoter_2', types=[sbol3.SBO_DNA], roles=[sbol3.SO_PROMOTER])

for obj in [pro1, pro2]:
    doc.add(obj) 
```

Create a `Collection` object and add the `TopLevels` IRI to it

```python
collection = sbol3.Collection(identity='example_collection', 
                              members=[pro1.identity, pro2.identity])
doc.add(collection)
```

Validate the document

```python
for obj in doc.objects:
    print(obj.identity)
report_sbol3 = doc.validate()
print(len(report_sbol3))
```

Save the document to an SBOL file

```python
doc.write('../CreatedSBOL3Objects/collection_example.xml')
```

```python
# Write document as sorted N-Triples (good for version control)
doc.write('../CreatedSBOL3Objects/collection_example.nt', sbol3.SORTED_NTRIPLES)
```
