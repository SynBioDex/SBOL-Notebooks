# Creating a Attachment

In the following tutorial, we will be creating a `Attachment` object.

`Attachment` objects have the following properties:
- `source` REQUIRED and MUST contain a IRI reference to the source file
- `format` OPTIONAL and MAY contain a IRI that specifies the format of the attached file
- `size` OPTIONAL and MAY contain a long indicating the file size in bytes
- `hash` OPTIONAL and MAY contain a hash value for the file contents represented as a hexadecimal digest.
- `hashAlgorithm` OPTIONAL and MAY contain the name of the hash algorithm used to generate the value of the hash property.

The `Attachment` class serve as a general container for data files, especially experimental data files. It provides a means for linking files and metadata to SBOL designs

For more information on the `Attachment` class and its properties, check out page 39 of the SBOL 3.1 specifications which can be found at the following [link](https://sbolstandard.org/docs/SBOL3.1.0.pdf).

Import the module

```python
import sbol3
```

Create the document and set the namespace

```python
doc = sbol3.Document()
sbol3.set_namespace('https://github.com/SynBioDex/SBOL-Notebooks')
```

Create an `Attachment` object pointing to a file

```python
attachment_example = sbol3.Attachment(
    identity='attachment_1',
    source='../data/experimental_data_1.csv')
doc.add(attachment_example)
```

Validate the document

```python
for obj in doc.objects:
    print(obj.identity)
report_sbol3 = doc.validate()
print(len(report_sbol3))
```

Save the document to an SBOL file

```python
doc.write('../CreatedSBOL3Objects/attachment_example.xml')
```

```python
# Write document as sorted N-Triples (good for version control)
doc.write('../CreatedSBOL3Objects/attachment_example.nt', sbol3.SORTED_NTRIPLES)
```
