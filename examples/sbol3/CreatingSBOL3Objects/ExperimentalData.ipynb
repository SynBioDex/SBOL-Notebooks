# Creating an Experimental Data

In the following tutorial, we will be creating a `ExperimentData` object.

`ExperimentData` objects have the following properties:
- `attachments` Zero or more file locations encoding data.


The `ExperimentData` class groups together a set of `TopLevel` objects that have something in common.

For more information on the `ExperimentData` class and its properties, check out page 39 of the SBOL 3.1 specifications which can be found at the following [link](https://sbolstandard.org/docs/SBOL3.1.0.pdf).

Import the module

```python
import sbol3
```

Create the document and set the namespace

```python
doc = sbol3.Document()
sbol3.set_namespace('https://github.com/SynBioDex/SBOL-Notebooks')
```

Create an `Attachment` object pointing to a file

```python
attachment_example = sbol3.Attachment(
    identity='attachment_1',
    source='../data/experimental_data_1.csv')
doc.add(attachment_example)
```

Create an `ExperimentalData` object and add an `Attachment` to it

```python
experimental_data_example = sbol3.ExperimentalData(
    identity='exp_data_1',
    attachments=[attachment_example])
doc.add(experimental_data_example)
```

Validate the document

```python
for obj in doc.objects:
    print(obj.identity)
report_sbol3 = doc.validate()
print(len(report_sbol3))
```

Save the document to an SBOL file

```python
doc.write('../CreatedSBOL3Objects/experimentaldata_example.xml')
```

```python
# Write document as sorted N-Triples (good for version control)
doc.write('../CreatedSBOL3Objects/experimentaldata_example.nt', sbol3.SORTED_NTRIPLES)
```
